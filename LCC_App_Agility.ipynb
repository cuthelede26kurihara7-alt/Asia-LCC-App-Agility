```python
def classify_airasia(note):
    note = note.lower()

    if "levelled up" in note or "experience" in note or "sleek" in note:
        return "Customer Experience"
    
    return "Maintenance"
```

```python
all_apps.loc[
    all_apps["airline"]=="AirAsia",
    "category"
] = all_apps.loc[
    all_apps["airline"]=="AirAsia",
    "release_note"
].apply(classify_airasia)
```

```python
all_apps[
    all_apps["airline"]=="AirAsia"
]["category"].value_counts()
```

```python
category_mix = (
    all_apps
    .groupby(["airline","category"])
    .size()
    .reset_index(name="count")
)

category_ratio = (
    category_mix
    .assign(
        ratio=lambda x: 
        x["count"] / x.groupby("airline")["count"].transform("sum") * 100
    )
)

category_ratio
```

```python
pivot = category_ratio.pivot(
    index="airline",
    columns="category",
    values="ratio"
).fillna(0)

pivot
```

```python
pivot.to_csv(
    "airline_app_agility_category_pivot.csv"
)
```
